# Target Encoding

```python
print(y.unique())
```

```python
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()

y_encoded = label_encoder.fit_transform(y)

print("Encoded classes:")
for original, encoded in zip(label_encoder.classes_, range(len(label_encoder.classes_))):
    print(original, "->", encoded)
```

# Train/Test Split

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y_encoded,
    test_size=0.20,
    random_state=42,
    stratify=y_encoded
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

# Encoding + Scaling Pipeline

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numerical_features),
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_features)
    ]
)
```

```python
X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

print("Processed training shape:", X_train_processed.shape)
print("Processed testing shape:", X_test_processed.shape)
```

# **Model-1: Decision Tree training**

```python
from sklearn.tree import DecisionTreeClassifier

dt_model = DecisionTreeClassifier(
    random_state=42
)
```

```python
dt_model.fit(X_train_processed, y_train)
```

```python
y_pred_dt = dt_model.predict(X_test_processed)
```

Decision Tree Evaluation

```python
from sklearn.metrics import accuracy_score

dt_accuracy = accuracy_score(y_test, y_pred_dt)

print("Decision Tree Accuracy:", dt_accuracy)
```

```python
from sklearn.metrics import classification_report

print(classification_report(
    y_test,
    y_pred_dt,
    target_names=label_encoder.classes_
))
```

Confusion Matrix

```python
from sklearn.metrics import confusion_matrix

cm_dt = confusion_matrix(y_test, y_pred_dt)

plt.figure(figsize=(8, 6))

sns.heatmap(
    cm_dt,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=label_encoder.classes_,
    yticklabels=label_encoder.classes_
)

plt.title('Decision Tree Confusion Matrix')
plt.xlabel('Predicted Label')
plt.ylabel('Actual Label')
plt.xticks(rotation=45)
plt.yticks(rotation=0)
plt.show()
```

ROC-AUC

```python
from sklearn.metrics import roc_auc_score
from sklearn.preprocessing import label_binarize

y_test_binary = label_binarize(
    y_test,
    classes=range(len(label_encoder.classes_))
)

y_prob_dt = dt_model.predict_proba(X_test_processed)

dt_auc = roc_auc_score(
    y_test_binary,
    y_prob_dt,
    multi_class='ovr',
    average='weighted'
)

print("Decision Tree ROC-AUC:", dt_auc)
```

ROC Curve

```python
from sklearn.metrics import roc_curve, auc

plt.figure(figsize=(9, 7))

for i, class_name in enumerate(label_encoder.classes_):
    fpr, tpr, _ = roc_curve(
        y_test_binary[:, i],
        y_prob_dt[:, i]
    )

    roc_auc = auc(fpr, tpr)

    plt.plot(
        fpr,
        tpr,
        label=f'{class_name} (AUC = {roc_auc:.2f})'
    )

plt.plot([0, 1], [0, 1], linestyle='--')

plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('Decision Tree ROC Curve')
plt.legend()
plt.show()
```

# **Model-5: K-Means Clustering**

```python
from sklearn.cluster import KMeans
```

Elobow Method

```python
inertia = []

K_range = range(2, 11)

for k in K_range:
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    kmeans.fit(X_train_processed)
    inertia.append(kmeans.inertia_)
```

```python
plt.figure(figsize=(8, 5))

plt.plot(K_range, inertia, marker='o')

plt.title('Elbow Method for Optimal K')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Inertia')

plt.show()
```

```python
kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

clusters = kmeans.fit_predict(X_train_processed)
```

Cluster **Distribution**

```python
optimal_k = 3

kmeans = KMeans(
    n_clusters=optimal_k,
    random_state=42,
    n_init=10
)

clusters = kmeans.fit_predict(X_train_processed)

print("Cluster Distribution:")
print(pd.Series(clusters).value_counts().sort_index())
```

Cluster **Visualization**

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2, random_state=42)

X_train_pca = pca.fit_transform(X_train_processed)

print("Explained Variance Ratio:")
print(pca.explained_variance_ratio_)
```

```python
plt.figure(figsize=(9, 6))

sns.scatterplot(
    x=X_train_pca[:, 0],
    y=X_train_pca[:, 1],
    hue=clusters,
    palette='viridis'
)

plt.title('K-Means Cluster Visualization')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.legend(title='Cluster')

plt.show()
```

**Cluster vs Depression Label**

```python
cluster_analysis = pd.crosstab(
    clusters,
    y_train,
    normalize='index'
) * 100

cluster_analysis.columns = label_encoder.inverse_transform(
    cluster_analysis.columns.astype(int)
)

print(cluster_analysis.round(2))
```

```python
cluster_analysis.plot(
    kind='bar',
    figsize=(12, 6)
)

plt.title('Depression Distribution Across K-Means Clusters')
plt.xlabel('Cluster')
plt.ylabel('Percentage')
plt.xticks(rotation=0)
plt.legend(title='Depression Level')

plt.show()
```

## **Final Model Comparison**

```python
from sklearn.metrics import precision_score, recall_score, f1_score

models = {
    'Decision Tree': y_pred_dt,
    'Logistic Regression': y_pred_lr,
    'Neural Network': y_pred_nn,
    'KNN': y_pred_knn
}

results = []

for model_name, predictions in models.items():
    results.append({
        'Model': model_name,
        'Accuracy': accuracy_score(y_test, predictions),
        'Precision': precision_score(
            y_test, predictions, average='weighted', zero_division=0
        ),
        'Recall': recall_score(
            y_test, predictions, average='weighted', zero_division=0
        ),
        'F1 Score': f1_score(
            y_test, predictions, average='weighted', zero_division=0
        )
    })

results_df = pd.DataFrame(results)

results_df
```

**ROC-AUC Comparison**

```python
auc_results = pd.DataFrame({
    'Model': [
        'Decision Tree',
        'Logistic Regression',
        'Neural Network',
        'KNN'
    ],
    'ROC-AUC': [
        dt_auc,
        lr_auc,
        nn_auc,
        knn_auc
    ]
})

auc_results
```

```python
final_results = results_df.merge(
    auc_results,
    on='Model'
)

final_results
```

 **Accuracy Comparison Chart**

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=final_results,
    x='Model',
    y='Accuracy'
)

plt.title('Accuracy Comparison of Classification Models')
plt.xlabel('Model')
plt.ylabel('Accuracy')
plt.ylim(0, 1)

plt.xticks(rotation=20)
plt.show()
```

**Precision, Recall & F1 Comparison**

```python
comparison_metrics = final_results.melt(
    id_vars='Model',
    value_vars=['Precision', 'Recall', 'F1 Score'],
    var_name='Metric',
    value_name='Score'
)

plt.figure(figsize=(12, 6))

sns.barplot(
    data=comparison_metrics,
    x='Model',
    y='Score',
    hue='Metric'
)

plt.title('Precision, Recall and F1 Score Comparison')
plt.xlabel('Model')
plt.ylabel('Score')
plt.ylim(0, 1)

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=final_results,
    x='Model',
    y='ROC-AUC'
)

plt.title('ROC-AUC Comparison of Classification Models')
plt.xlabel('Model')
plt.ylabel('ROC-AUC')
plt.ylim(0, 1)

plt.xticks(rotation=20)
plt.show()
```

```python
probabilities = {
    'Decision Tree': y_prob_dt,
    'Logistic Regression': y_prob_lr,
    'Neural Network': y_prob_nn,
    'KNN': y_prob_knn
}

plt.figure(figsize=(10, 8))

for model_name, probabilities_model in probabilities.items():

    fpr_all = []
    tpr_all = []

    for i in range(len(label_encoder.classes_)):
        fpr, tpr, _ = roc_curve(
            y_test_binary[:, i],
            probabilities_model[:, i]
        )

        fpr_all.extend(fpr)
        tpr_all.extend(tpr)

    # Sort points for visualization
    sorted_indices = np.argsort(fpr_all)

    plt.plot(
        np.array(fpr_all)[sorted_indices],
        np.array(tpr_all)[sorted_indices],
        label=f'{model_name} (AUC = {dict(zip(auc_results["Model"], auc_results["ROC-AUC"]))[model_name]:.3f})'
    )

plt.plot([0, 1], [0, 1], linestyle='--')

plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve Comparison')
plt.legend()
plt.show()
```
